```python
def pattern9(n):
  space=2*(n-1);
  for i in range(1,n+1):
    for j in range(1,i+1):
      print(j,end=' ')
    for j in range(1,space+1):
      print(' ', end=' ')
    for j in range(i,0,-1):
      print(j,end=' ')
    print('\n')
    space-=2

pattern9(4)
```

```python
def pattern14(n):
  char=ord('A')
  for i in range(n):
    break_point=(2*i+1)//2
    char=ord('A')
    # print(break_point)
    for j in range(n-i-1):
      print(' ',end=' ')
    for j in range(2*i+1):
      print(chr(char), end=' ')
      if j<break_point:
        char+=1
      else:
        char-=1
    for j in range(n-i-1):
      print(' ',end=' ')
    print('\n')

pattern14(5)
```

```python
def pattern18(n):
  inis=2*n-2
  for i in range(2*n-1):
    stars=i+1
    if (i>=n):
      stars=2*n-i-1
    for j in range(stars):
      print('*',end=' ')
    for j in range(inis):
      print(' ', end=' ')
    for j in range(stars):
      print('*',end=' ')
    print()
    if (i<n-1):
      inis-=2
    else:
      inis+=2

pattern18(5)
```

```python
def pattern18(n):
    inis = 2 * n - 2

    for i in range(2 * n - 1):

        stars = i + 1

        if i >= n:
            stars = 2 * n - i - 1

        # Left stars
        for j in range(stars):
            print('*', end=' ')

        # Middle spaces
        for j in range(inis):
            print(' ', end=' ')

        # Right stars
        for j in range(stars):
            print('*', end=' ')

        print()

        if i < n - 1:
            inis -= 2
        else:
            inis += 2


pattern18(5)
```
